# Partie 9 : Gestion des valeurs manquantes

```python
#valeur nullle par colone
nulCol=df.isna().sum()
nulCol
```

```python
total=df.isna().sum().sum()
total
```

```python
proption_valur_nulle=(nulCol/total)*100
proption_valur_nulle
```

```python
lignes_manquantes = df[df.isna().any(axis=1)]
lignes_manquantes
```

```python
df["temperature"] = df["temperature"].fillna(df["temperature"].mean())
df["humidite"] = df["humidite"].fillna(df["humidite"].mean())
```

```python
df["consommation"] = df["consommation"].fillna(df["consommation"].median())
```

```python
df["etat"] = df["etat"].fillna("INCONNU")
```
