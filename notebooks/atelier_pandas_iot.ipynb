# Partie 1 – Series

```python
import pandas as pd
```

```python
temperatures = pd.Series([22.5, 24.0, 21.8, 25.3])
print(temperatures)
```

```python
print(temperatures.index)
```

```python
print(temperatures.iloc[0])
```

```python
print(temperatures.iloc[-1])
```

```python
heures = ["12h", "13h", "14h", "15h"]
temp_series1 = pd.Series([22.5, 24.0, 21.8, 25.3], index=heures)

temp_dict = {"12h": 22.5, "13h": 24.0, "14h": 21.8, "15h": 25.3}
temp_series2 = pd.Series(temp_dict)

print("Méthode 1 :\n", temp_series1)
print("\nMéthode 2 :\n", temp_series2)
```

```python
print(temp_series1[["12h", "15h"]])
```
